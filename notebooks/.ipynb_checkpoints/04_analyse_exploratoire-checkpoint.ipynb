# étape 4 - analyse exploratoire des tirs (SMOE)

ce notebook décrit les données nettoyées ('data/processed/shots_features.parquet') avant la modélisation. chaque figure est sauvegardée dans 'figures/' pour le rapport.

1. vue d'ensemble
2. réussite et valeur selon la distance
3. réussite par zone et par famille de tirs
4. carte de tirs de la ligue
5. variables de fatigue (brutes et ajustées pour la zone)
6. corrélations entre les variables
7. multicolinéarité (VIF)

```python
from pathlib import Path
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
import seaborn as sns

# se placer à la racine du projet si le notebook est dans notebooks/
if Path.cwd().name == "notebooks":
    os.chdir("..")

DATA_FILE = Path("data/processed/shots_features.parquet")
FIG_DIR = Path("figures")
FIG_DIR.mkdir(exist_ok=True)

# palette et style (sobres, lisibles à l'impression)
BLEU = "#257de0"
ORANGE = "#f47230"
ROUGE = "#ec4647"
GRIS_MILIEU = "#efefef"
TEXTE = "#191919"
TEXTE_2 = "#666666"
GRILLE = "#e5e5e5"

plt.rcParams.update({
    "figure.dpi": 110,
    "savefig.dpi": 200,
    "savefig.bbox": "tight",
    "font.size": 10,
    "axes.titlesize": 12,
    "axes.titleweight": "bold",
    "axes.titlelocation": "left",
    "axes.edgecolor": GRILLE,
    "axes.labelcolor": TEXTE_2,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": GRILLE,
    "grid.linewidth": 0.8,
    "xtick.color": TEXTE_2,
    "ytick.color": TEXTE_2,
    "text.color": TEXTE,
    "legend.frameon": False,
})

# divergent : blue (sous la moyenne) -> gris -> rouge (au-dessus)
CMAP_DIV = LinearSegmentedColormap.from_list("div", [BLEU, GRIS_MILIEU, ROUGE])


def sauver(fig, nom):
    """sauvegarde une figure dans figures/ (png haute résolution)"""
    chemin = FIG_DIR / f"{nom}.png"
    fig.savefig(chemin)
    print(f"figure sauvegardée : {chemin}")


def taux_avec_ic(df, colonne, cible="SHOT_MADE_FLAG"):
    """taux de réussite par groupe avec intervalle de confiance à 95%"""
    g = df.groupby(colonne, observed=True)[cible].agg(n="size", taux="mean").reset_index()
    g["ic95"] = 1.96 * np.sqrt(g["taux"] * (1 - g["taux"]) / g["n"])
    return g
```

```python
# chargement des données
tirs = pd.read_parquet(DATA_FILE)
MOY_LIGUE = tirs["SHOT_MADE_FLAG"].mean()
tirs["POINTS"] = tirs["SHOT_MADE_FLAG"] * tirs["VALEUR_TIR"]
print(f"{len(tirs):,} tirs, {tirs.shape[1]} colonnes")
print(f"Taux de réussite de la ligue : {MOY_LIGUE:.3f}")
print(f"Points par tir de la ligue   : {tirs['POINTS'].mean():.3f}")
```

## 1. vue d'ensemble

répartition de la cible et statistiques descriptives des variables numériques.

```python
print("répartition de la cible (SHOT_MADE_FLAG) :")
print(tirs["SHOT_MADE_FLAG"].value_counts(normalize=True).rename({0: "raté", 1: "réussi"}).round(3))
print("\ntirs à 2 et à 3 points :")
print(tirs.groupby("VALEUR_TIR")["SHOT_MADE_FLAG"].agg(tirs="size", reussite="mean").round(3))

NUMERIQUES = ["DIST_CALC", "ANGLE_TIR", "LOC_X", "LOC_Y", "PERIOD", "TEMPS_RESTANT",
              "TEMPS_ECOULE", "MINUTES_JOUEES", "JOURS_REPOS", "MATCHS_7J"]
tirs[NUMERIQUES].describe().T.round(2)
```

## 2. réussite et valeur selon la distance

la distance est la variable qu'on s'attend à voir dominer. le deuxième panneau montre les **points par tir** (taux x valeur) : c'est exactement ce que le xPts cherche à prédire. les lignes verticales marquent la ligne de 3 points (22 pi dans les coins, 23,75 pi ailleurs).

```python
d = tirs[tirs["DIST_CALC"] < 36].copy()
d["DIST_PI"] = d["DIST_CALC"].astype(int)
par_dist = d.groupby("DIST_PI").agg(n=("SHOT_MADE_FLAG", "size"),
                                    taux=("SHOT_MADE_FLAG", "mean"),
                                    points=("POINTS", "mean")).reset_index()
par_dist = par_dist[par_dist["n"] >= 100]   # retirer les distances trop rares

fig, axes = plt.subplots(3, 1, figsize=(8, 9), sharex=True,
                         gridspec_kw={"height_ratios": [3, 3, 2]})

axes[0].plot(par_dist["DIST_PI"], par_dist["taux"], color=BLEU, lw=2, marker="o", ms=4)
axes[0].axhline(MOY_LIGUE, color=TEXTE_2, lw=1, ls="--")
axes[0].text(par_dist["DIST_PI"].max(), MOY_LIGUE, f" moyenne {MOY_LIGUE:.1%}",
          va="bottom", ha="right", color=TEXTE_2, fontsize=9)
axes[0].set_ylabel("taux de réussite")
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
axes[0].set_title("taux de réussite selon la distance")

axes[1].plot(par_dist["DIST_PI"], par_dist["points"], color=BLEU, lw=2, marker="o", ms=4)
axes[1].set_ylabel("points par tir")
axes[1].set_title("points par tir selon la distance")

axes[2].bar(par_dist["DIST_PI"], par_dist["n"], color=BLEU, width=0.8)
axes[2].set_ylabel("nombre de tirs")
axes[2].set_xlabel("distance au panier (pi)")
axes[2].set_title("volume de tirs")
axes[2].yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v/1000:.0f}k"))

for ax in axes:
    for x in (22, 23.75):
        ax.axvline(x, color=TEXTE_2, lw=0.8, ls=":")
fig.tight_layout()
sauver(fig, "01_distance")
plt.show()
```

**à noter pour le rapport :** forme de la courbe (chute rapide après 3-4 pi, plateau à mi-distance), saut des points par tir à la ligne de 3, et creux du volume à mi-distance (le "désert" de la nba moderne).

## 3. réussite par zone et par famille de tirs

la barre d'erreur est l'intervalle de confiance à 95% ; 'n' est le nombre de tirs.

```python
def barres_taux(df, colonne, titre, nom_fig):
    g = taux_avec_ic(df, colonne).sort_values("taux")
    fig, ax = plt.subplots(figsize=(8, 0.45 * len(g) + 1.2))
    ax.barh(g[colonne].astype(str), g["taux"], xerr=g["ic95"], color=BLEU,
            height=0.6, error_kw={"ecolor": TEXTE_2, "lw": 1, "capsize": 3})
    ax.axvline(MOY_LIGUE, color=TEXTE_2, lw=1, ls="--")
    for i, (t, n) in enumerate(zip(g["taux"], g["n"])):
        ax.text(t + g["ic95"].max() + 0.01, i, f"{t:.1%}  (n = {n:,})",
                va="center", fontsize=9, color=TEXTE)
    ax.set_xlim(0, min(1, g["taux"].max() + 0.25))
    ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
    ax.set_xlabel("taux de réussite")
    ax.grid(axis="y", visible=False)
    ax.set_title(titre)
    fig.tight_layout()
    sauver(fig, nom_fig)
    plt.show()

barres_taux(tirs, "SHOT_ZONE_BASIC", "Taux de réussite par zone", "02_zones")
barres_taux(tirs, "FAMILLE_TIR", "taux de réussite par famille de tirs", "03_familles")
```

## 4. carte de tirs de la ligue

chaque hexagone montre l'écart entre le taux de réussite à cet endroit et la moyenne de la ligue : **rouge = mieux que la moyenne, blue = moins bien**. les hexagones de moins de 30 tirs sont masqués.

```python
def dessiner_terrain(ax, couleur="#999999", lw=1):
    """demi-terrain nba, coordonnées en dixièmes de pied, panier à l'origine."""
    elements = [
        patches.Circle((0, 0), radius=7.5, lw=lw, color=couleur, fill=False),         # anneau (basket)
        patches.Rectangle((-30, -10), 60, 0, lw=lw, color=couleur),                   # planche (floor)
        patches.Rectangle((-80, -47.5), 160, 190, lw=lw, color=couleur, fill=False),  # raquette (paint)
        patches.Arc((0, 142.5), 120, 120, theta1=0, theta2=180, lw=lw, color=couleur),
        patches.Arc((0, 0), 80, 80, theta1=0, theta2=180, lw=lw, color=couleur),      # zone restreinte (restricted area)
        patches.Rectangle((-220, -47.5), 0, 140, lw=lw, color=couleur),               # coins à 3 points (3 point corners)
        patches.Rectangle((220, -47.5), 0, 140, lw=lw, color=couleur),
        patches.Arc((0, 0), 475, 475, theta1=22, theta2=158, lw=lw, color=couleur),   # arc à 3 points (3 point arc)
        patches.Rectangle((-250, -47.5), 500, 470, lw=lw, color=couleur, fill=False),
    ]
    for e in elements:
        ax.add_patch(e)
    ax.set_xlim(-250, 250)
    ax.set_ylim(-47.5, 422.5)
    ax.set_aspect("equal")
    ax.axis("off")


fig, ax = plt.subplots(figsize=(8, 7.5))
hb = ax.hexbin(tirs["LOC_X"], tirs["LOC_Y"], C=tirs["SHOT_MADE_FLAG"] - MOY_LIGUE,
               reduce_C_function=np.mean, gridsize=40, mincnt=30,
               extent=(-250, 250, -47.5, 422.5), cmap=CMAP_DIV,
               norm=TwoSlopeNorm(vmin=-0.25, vcenter=0, vmax=0.25),
               edgecolors="white", linewidths=0.3)
dessiner_terrain(ax)
cb = fig.colorbar(hb, ax=ax, shrink=0.6, pad=0.02)
cb.set_label("écart au taux moyen de la ligue")
cb.ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:+.0%}"))
cb.outline.set_visible(False)
ax.set_title(f"réussite par position, écart à la moyenne de la ligue ({MOY_LIGUE:.1%})")
sauver(fig, "04_carte_tirs_ligue")
plt.show()
```

## 5. variables de fatigue

deux lectures pour chaque variable :

- **brute** : taux de réussite tel quel.
- **ajustée pour la zone** : on soustrait à chaque tir, le taux moyen de sa zone ('SHOT_ZONE_BASIC' x 'SHOT_ZONE_RANGE'), puis on fait la moyenne de ces écarts.
ce n'est **pas** le résultat final : seul le modèle isole vraiment l'effet de la fatigue. les barres d'erreur sont des intervalles de confiances à 95%.

```python
# écart de chaque tir au taux moyen de sa zone (ajustement simple, à titre exploratoire)
zone = tirs["SHOT_ZONE_BASIC"] + " | " + tirs["SHOT_ZONE_RANGE"]
tirs["ECART_ZONE"] = tirs["SHOT_MADE_FLAG"] - tirs.groupby(zone)["SHOT_MADE_FLAG"].transform("mean")

tirs["MINUTES_TRANCHE"] = pd.cut(tirs["MINUTES_JOUEES"], bins=[0, 6, 12, 18, 24, 30, 36, 60],
                                 labels=["0-6", "6-12", "12-18", "18-24", "24-30", "30-36", "36+"],
                                 include_lowest=True)
tirs["MATCHS_7J_CAT"] = tirs["MATCHS_7J"].clip(upper=4).astype(str).replace({"4": "4+"})
tirs["BACK_TO_BACK_LBL"] = tirs["BACK_TO_BACK"].map({0: "non", 1: "oui"})

VARIABLES_FATIGUE = [
    ("MINUTES_TRANCHE", "minutes jouées avant le tir"),
    ("JOURS_REPOS_CAT", "jours de repos"),
    ("BACK_TO_BACK_LBL", "back-to_back"),
    ("MATCHS_7J_CAT", "matchs dans les 7 jours précédents"),
]


def resume_fatigue(colonne):
    g = tirs.groupby(colonne, observed=True).agg(
        n=("SHOT_MADE_FLAG", "size"),
        taux=("SHOT_MADE_FLAG", "mean"),
        ecart=("ECART_ZONE", "mean"),
        ecart_sd=("ECART_ZONE", "std"),
    ).reset_index()
    g["ic_taux"] = 1.96 * np.sqrt(g["taux"] * (1 - g["taux"]) / g["n"])
    g["ic_ecart"] = 1.96 * g["ecart_sd"] / np.sqrt(g["n"])
    return g


fig, axes = plt.subplots(2, 4, figsize=(14, 6.5), sharey="row")
for j, (col, titre) in enumerate(VARIABLES_FATIGUE):
    g = resume_fatigue(col)
    x = np.arange(len(g))

    ax = axes[0, j]
    ax.errorbar(x, g["taux"], yerr=g["ic_taux"], color=BLEU, marker="o", ms=6, lw=2,
                capsize=3, ecolor=TEXTE_2, elinewidth=1)
    ax.axhline(MOY_LIGUE, color=TEXTE_2, lw=1, ls="--")
    ax.set_title(titre, fontsize=10)
    ax.set_xticks(x, g[col].astype(str))
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.1%}"))

    ax = axes[1, j]
    ax.errorbar(x, g["ecart"], yerr=g["ic_ecart"], color=ORANGE, marker="o", ms=6, lw=2,
                capsize=3, ecolor=TEXTE_2, elinewidth=1)
    ax.axhline(0, color=TEXTE_2, lw=1, ls="--")
    ax.set_xticks(x, g[col].astype(str))
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:+.1%}"))

axes[0, 0].set_ylabel("taux de réussite brut")
axes[1, 0].set_ylabel("écart au taux de la zone")
fig.suptitle("réussite selon les variables de fatigue - brute (haut) et ajustée pour la zone (bas)",
             x=0.01, ha="left", fontweight="bold")
fig.tight_layout()
sauver(fig, "05_fatigue")
plt.show()

for col, titre in VARIABLES_FATIGUE:
    print(f"\n{titre}")
    print(resume_fatigue(col)[[col, "n", "taux", "ecart"]].round(4).to_string(index=False))
```

**à noter pour le rapport :** si l'écart ajusté reste dans l'intervalle de confiance de zéro, l'effait de la variable est faible ou nul une fois la zone prise en compte.

## 6. corrélations entre les variables numériques

on cherche les **redondances** : des variables qui portent presque la même information.

```python
VARS_MODELE = ["DIST_CALC", "ANGLE_TIR", "LOC_X", "LOC_Y", "VALEUR_TIR",
               "PERIOD", "TEMPS_RESTANT", "TEMPS_ECOULE",
               "MINUTES_JOUEES", "JOURS_REPOS", "BACK_TO_BACK", "MATCHS_7J", "DEBUT_HISTORIQUE"]

corr = tirs[VARS_MODELE].corr()
masque = np.triu(np.ones_like(corr, dtype=bool), k=1)

fig, ax = plt.subplots(figsize=(10, 8.5))
sns.heatmap(corr, mask=masque, cmap=CMAP_DIV, vmin=-1, vmax=1, center=0,
            annot=True, fmt=".2f", annot_kws={"size": 8}, square=True,
            linewidths=1, linecolor="white", cbar_kws={"shrink": 0.6, "label": "corrélation des variables"}, ax=ax)
ax.grid(False)
ax.set_title("matrice de corrélation des variables numériques")
sauver(fig, "06_ correlations")
plt.show()

paires = (corr.where(~np.tril(np.ones_like(corr, dtype=bool)))
          .stack().rename("corr").reset_index())
paires["abs"] = paires["corr"].abs()
print("paires les plus corrélées (|r| > 0,5) :")
print(paires[paires["abs"] >= 0.5].sort_values("abs", ascending=False)
      [["level_0", "level_1", "corr"]].round(3).to_string(index=False))
```

## multicolinéarité (vif)

on régresse chaque variable sur toutes les autres : vif = 1 / (1-R²).

- vif < 5 : correct
- 5 à 10 : à surveiller
- \> 10 :redondance forte - à retirer **pour la régression logistique** (les arbres n'en souffre pas)

```python
def calculer_vif(df, colonnes):
    """VIF_J = 1 / (1 - R²_J), où R²_J vient de la régression de la variable j sure les autres."""
    X = df[colonnes].astype(float)
    X = ((X - X.mean()) / X.std()).to_numpy()
    vifs = []
    for j in range(X.shape[1]):
        y = X[:, j]
        autres = np.delete(X, j, axis=1)
        coef, *_ = np.linalg.lstsq(autres, y, rcond=None)
        r2 = 1 - ((y - autres @ coef) ** 2).sum() / (y**2).sum()
        vifs.append(np.inf if r2 > 0.9999 else 1 / (1 - r2))
    return (pd.DataFrame({"variable": colonnes, "VIF": vifs})
            .sort_values("VIF", ascending=False).reset_index(drop=True))


def niveau(v):
    return "fort (> 10)" if v > 10 else ("à surveiller" if v > 5 else "correct")


vif_tout = calculer_vif(tirs, VARS_MODELE)
vif_tout["niveau"] = vif_tout["VIF"].map(niveau)
print("vif - toutes les variables numériques :")
print(vif_tout.round(2).to_string(index=False))
```

on s'attend à des vif élevés pour 'TEMPS_ECOULE' / 'PERIOD' / 'TEMPS_RESTANT', 'DIST_CALC' / 'LOC_Y' / 'VALEUR_TIR', et 'JOURS_REPOS' / 'BACK_TO_BACK'.

essai d'un **jeu réduit** pour la régression logistique :

```python
VARS_LOGISTIQUE = ["DIST_CALC", "ANGLE_TIR", "VALEUR_TIR", "TEMPS_RESTANT",
                   "MINUTES_JOUEES", "JOURS_REPOS", "MATCHS_7J", "DEBUT_HISTORIQUE"]

vif_reduit = calculer_vif(tirs, VARS_LOGISTIQUE)
vif_reduit["niveau"] = vif_reduit["VIF"].map(niveau)
print("vif - jeu réduit pour la régression logistique :")
print(vif_reduit.round(2).to_string(index=False))
```
